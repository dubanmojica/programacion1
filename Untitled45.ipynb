```python
import re

class CorreoInvalidoError(Exception):
    """Excepción cuando el correo no cumple con la forma institucional."""
    pass
def validar_correo(correo):
    patron = r"^[a-z0-9_]+@universitario\.edu\.co$"
    if not re.fullmatch(patron, correo):
        raise CorreoInvalidoError(
            "Correo inválido"
        )
    return True
def main():
    intentos = 0
    max_intentos = 3
    while True:
        try:
            correo = input("Ingrese su correo institucional: ")
            validar_correo(correo)
            print("Correo válido")
            break

        except CorreoInvalidoError as e:
            print("Error:", e)
            intentos += 1
            if intentos >= max_intentos:
                print("Acceso documento: Acceso bloqueado por intentos fallidos.")
                break

        except Exception as e:
            print("Error inesperado:", e)
            intentos += 1
            if intentos >= max_intentos:
                print("Acceso bloqueado por intentos fallidos.")
                break
if __name__ == "__main__":
    main()
```
