```python
import sys

try:
    # 1. Primer valor: Si falla aquí, salta directo al except y no pide nada más
    num1 = float(input("numero1: "))

    # 2. Segundo valor: Solo se pide si num1 fue exitoso
    num2 = float(input("numero2: "))

    # 3. Operación
    op = input("operacion (+, -, *, /, %): ")

    if op == "+":
        resultado = num1 + num2
    elif op == "-":
        resultado = num1 - num2
    elif op == "*":
        resultado = num1 * num2
    elif op == "/":
        resultado = num1 / num2
    elif op == "%":
        resultado = num1 % num2
    else:
        # Forzamos el error si la operación no existe
        raise ValueError("operacion no valida")

    print("resultado:", resultado)

except ValueError as e:
    print("Error de valor:", e)
    sys.exit() # Cierre inmediato y formal del programa

except ZeroDivisionError as e:
    print("No se puede dividir entre cero:", e)
    sys.exit()

except Exception as e:
    print("Error inesperado:", e)
    sys.exit()
```
