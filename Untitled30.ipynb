```python
import numpy as np
import pandas as pd
np.random.seed(123)
```

```python
datos = np.random.normal(loc=4.0, scale=0.5, size=50)
```

```python
df = pd.DataFrame({"calificacion": datos})
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm
import numpy as np

plt.figure(figsize=(10, 6))
sns.histplot(df['calificacion'], kde=True, stat='density', bins=10, color='skyblue', label='Histograma + KDE')

# Calculate mean and standard deviation for the normal distribution
mu, std = df['calificacion'].mean(), df['calificacion'].std()

# Plot the PDF for the normal distribution
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = norm.pdf(x, mu, std)
plt.plot(x, p, 'k', linewidth=2, label='Normal PDF')

plt.title('Distribución de Calificaciones con Histograma, KDE y Normal PDF', fontsize=14)
plt.xlabel('Calificación', fontsize=12)
plt.ylabel('Densidad', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
from scipy.stats import shapiro
shapiro(df["calificacion"])
```

como el valor p es mayor que 0.05 no se rechaza la hipotesis de normalidad con confianza del 77%

```python
import numpy as np
import pandas as pd

np.random.seed(123)

grupo_A = np.random.normal(loc=4.0, scale=0.4, size=40)
grupo_B = np.random.normal(loc=4.3, scale=0.5, size=40)
```

```python
df = pd.DataFrame({
    "A": grupo_A,
    "B": grupo_B
})
```

```python
from scipy import stats
```

```python
from scipy.stats import shapiro
shapiro(df["A"])
```

```python
from scipy.stats import shapiro
shapiro(df["B"])
```

```python
df.describe()
```

```python
from scipy import stats
t_stat, p_val = stats.ttest_ind(df["A"], df["B"])

print("Estadístico t:", t_stat)
print("Valor p:", p_val)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm
import numpy as np

plt.figure(figsize=(10, 6))
sns.histplot(pd.concat([df['A'], df['B']]), kde=True, stat='density', bins=10, color='skyblue', label='Histograma + KDE')

# Calculate mean and standard deviation for the normal distribution
# Use the combined data for calculation
combined_data = pd.concat([df['A'], df['B']])
mu, std = combined_data.mean(), combined_data.std()

# Plot the PDF for the normal distribution
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = norm.pdf(x, mu, std)
plt.plot(x, p, 'k', linewidth=2, label='Normal PDF')

plt.title('Distribución de Calificaciones con Histograma, KDE y Normal PDF', fontsize=14)
plt.xlabel('Calificación', fontsize=12)
plt.ylabel('Densidad', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

se rechanz que las calificaciones de ambos grupos sean iguales con una confianza del 95 %
