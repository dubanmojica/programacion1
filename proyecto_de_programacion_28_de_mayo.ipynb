```python
dataf = pd.get_dummies(
    tabla_cocaina,
    columns=['dias_consumo', 'consumo_30d',  "consumo_12m"],
    prefix=['dias_consumo', 'consumo_30d',  "consumo_12m"]
)
dataf.head(4)
```

```python
import pandas as pd
dataf = pd.get_dummies(tabla_cocaina, columns=[
    'frecuencia_12m',
    'dias_consumo',
    'consumo_30d',
    'consumo_12m'
])
columnas_a_convertir = [
    'frecuencia_12m_1', 'frecuencia_12m_2',
    'dias_consumo_0.0', 'dias_consumo_2.0', 'dias_consumo_3.0', 'dias_consumo_4.0', 'dias_consumo_6.0', 'dias_consumo_10.0', 'dias_consumo_30.0',
    'consumo_30d_0.0', 'consumo_30d_1.0', 'consumo_30d_2.0',
    'consumo_12m_0.0', 'consumo_12m_1.0'
]
dataf[columnas_a_convertir] = dataf[columnas_a_convertir].astype(int)
dataf.head(4)
```

```python
from sklearn.model_selection import train_test_split
train_data, test_data = train_test_split(dataf, test_size=0.2, random_state=42)
```

```python
target_y_column = 'consumo_30d_0.0'
columns_to_drop_from_X = [col for col in dataf.columns if col.startswith('consumo_30d_')]
X_train = train_data.drop(columns=columns_to_drop_from_X, axis=1)
y_train = train_data[target_y_column]
X_test = test_data.drop(columns=columns_to_drop_from_X, axis=1)
y_test = test_data[target_y_column]
```

```python
y_test.head(3)
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
lda = LinearDiscriminantAnalysis()
lda.fit(X_train, y_train)
```

```python
y_pred = lda.predict(X_test) # Se hace prediccion sobre los datos de validacion para la variable explicativa
print("Predicciones:", y_pred)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

# Calcular matriz
cm = confusion_matrix(y_test, y_pred)

# Crear figura
plt.figure(figsize=(6,5))

# Dibujar heatmap bonito
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    cbar=True,
    linewidths=1,
    linecolor='white'
)

# Etiquetas y título
plt.title('Matriz de Confusión', fontsize=14, fontweight='bold')
plt.xlabel('Predicción')
plt.ylabel('Valor Real')

plt.tight_layout()
plt.show()
```

De acuerdo con la matriz de confusión, se encuentra que el algoritmo realiza predicciones altamente adecuadas para la clase de mayor frecuencia, logrando clasificar correctamente la totalidad de sus instancias.

```python
from sklearn.metrics import ConfusionMatrixDisplay
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot()
plt.title("Matriz de Confusión - LDA")
plt.show()
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```


El 80 % de los datos de la categoría 1 se predicen en esa misma categoría, de manera análoga el 100% de la categoría 0

```python
from sklearn.ensemble import RandomForestClassifier

modelo_rf = RandomForestClassifier(
    n_estimators=300,
    class_weight='balanced',
    random_state=42
)

modelo_rf.fit(X_train, y_train)

y_pred_rf = modelo_rf.predict(X_test)

print(classification_report(y_test, y_pred_rf))
```

```python
from sklearn.linear_model import LogisticRegression

modelo_log = LogisticRegression(
    max_iter=1000,
    class_weight='balanced'
)

modelo_log.fit(X_train, y_train)

y_pred_log = modelo_log.predict(X_test)

print(classification_report(y_test, y_pred_log))
```

El 80% de los datos que el primer algoritmo predice en la categoría 1 realmente pertenecen a esa clase, de manera análoga al 100% para la categoría 0. Al evaluar el rendimiento general, los modelos con la mejor capacidad de predicción son tanto el Random Forest como la Regresión Logística, logrando un acierto perfecto del 100% en la categoría 1 y manteniendo un 100% simétrico en la categoría 0
