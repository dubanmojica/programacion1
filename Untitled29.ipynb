```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm
import numpy as np

plt.figure(figsize=(10, 6))
sns.histplot(df['calificacion'], kde=True, stat='density', bins=10, color='skyblue', label='Histograma + KDE')

# Calculate mean and standard deviation for the normal distribution
mu, std = df['calificacion'].mean(), df['calificacion'].std()

# Plot the PDF for the normal distribution
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = norm.pdf(x, mu, std)
plt.plot(x, p, 'k', linewidth=2, label='Normal PDF')

plt.title('Distribución de Calificaciones con Histograma, KDE y Normal PDF', fontsize=14)
plt.xlabel('Calificación', fontsize=12)
plt.ylabel('Densidad', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd

np.random.seed(123)

grupo_A = np.random.normal(loc=4.0, scale=0.4, size=40)
grupo_B = np.random.normal(loc=4.3, scale=0.5, size=40)

```

```python
df = pd.DataFrame({
    "A": grupo_A,
    "B": grupo_B
})

```

```python
from scipy import stats
```

```python
from scipy import stats
t_stat, p_val = stats.ttest_ind(df["A"], df["B"])

print("Estadístico t:", t_stat)
print("Valor p:", p_val)
```

se rechanz que las calificaciones de ambos grupos sean iguales con una confianza del 95 %
